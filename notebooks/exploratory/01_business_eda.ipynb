# Business EDA

Schema-driven business exploratory analysis for the IBM Telco Customer Churn Status dataset.
Loads config, cleans the raw data via `PreprocessingPipeline`, then runs `BusinessEDA`.

```python
import pandas as pd

from churn_platform.utils.config import load_config
from churn_platform.preprocessing.pipeline import PreprocessingPipeline
from churn_platform.analysis.business_eda import BusinessEDA

config = load_config("config/config.yaml")
schema = config["schema"]
```

```python
raw_df = pd.read_excel(config["paths"]["raw_dataset_file"])

pipeline = PreprocessingPipeline(config=config["runtime"], schema=schema)
clean_df = pipeline.transform(raw_df)
clean_df.shape
```

```python
# Schema-driven: semantic column mapping is injected from config.
eda = BusinessEDA(clean_df, schema=schema.get("eda"))
```
